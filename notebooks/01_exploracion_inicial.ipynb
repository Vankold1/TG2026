## Exploración inicial de datos

Exploración de los subconjuntos de datos seleccionados del dataset LITNET-2020. El objetivo es comprender la estructura de cada archivo, identificar la cantidad de registros, variables disponibles, presencia de valores nulos y la distribución de las etiquetas de clasificación.

Flujos de ataques: Blaster Worm, Reaper Worm, Red Worm, HTTP flood, ICMP flood y UDP flood.

```python
import pandas as pd
import numpy as np
import os
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

```python
DATA_PATH = "../data/raw/"
files = [
    "BLASTER_WORM_v2.csv",
    "HTTP_FLOOD_v2.csv",
    "ICMP_FLOOD_v2.csv",
    "REAPER_WORM_v2.csv",
    "RED_WORM_v2.csv",
    "UDP_FLOOD_v2.csv"
]
files_paths = [os.path.join(DATA_PATH, f) for f in files]
files_paths
```

```python
column_names = [f"feature_{i+1}" for i in range(83)] + ["attack_t", "attack_a"]
```

```python
def explorar(path, nrows=100000):
    print("="*80)
    print(f"Archivo: {os.path.basename(path)}")
    
    df = pd.read_csv(path, header=None, names=column_names, nrows=nrows)
    
    print("\nShape (muestra):", df.shape)
    
    print("\nPrimeras filas:")
    display(df.head())
    
    print("\nTipos de datos:")
    print(df.dtypes.value_counts())
    
    print("\nValores nulos (%):")
    nulls = df.isnull().mean().sort_values(ascending=False)
    print(nulls.head(10))
    
    print("\nDistribución attack_a:")
    print(df["attack_a"].value_counts(dropna=False))
    print(df["attack_a"].value_counts(normalize=True, dropna=False))
    
    print("\nDistribución attack_t:")
    print(df["attack_t"].value_counts(dropna=False))
    
    print("\nMemoria estimada muestra (MB):")
    print(round(df.memory_usage(deep=True).sum() / 1024**2, 2))
    
    return df
```

```python
for path in files_paths:
    _ = explorar(path)
```

```python
resumen = []
for path in files_paths:
    df = pd.read_csv(path, header=None, names=column_names, nrows=100000)
    resumen.append({
        "archivo": os.path.basename(path),
        "filas_muestra": df.shape[0],
        "columnas": df.shape[1],
        "nulos_%_max_muestra": df.isnull().mean().max(),
        "attack_a_0_muestra": (df["attack_a"] == 0).sum(),
        "attack_a_1_muestra": (df["attack_a"] == 1).sum(),
        "attack_ratio_muestra": df["attack_a"].mean(),
        "attack_t_valores_muestra": df["attack_t"].unique().tolist()
    })
resumen_df = pd.DataFrame(resumen)
resumen_df
```

Nulos en todo el cvs

```python
def revisar_nulos_completo(path, chunksize=500000):
    total_filas = 0
    nulos = pd.Series(0, index=column_names, dtype="int64")
    for chunk in pd.read_csv(path, header=None, names=column_names, chunksize=chunksize):
        total_filas += len(chunk)
        nulos += chunk.isnull().sum()
    porcentaje_nulos = (nulos / total_filas) * 100
    return pd.DataFrame({
        "columna": column_names,
        "nulos": nulos.values,
        "porcentaje_nulos": porcentaje_nulos.values
    }).sort_values("porcentaje_nulos", ascending=False)
```

```python
nulos_globales = {}
for path in files_paths:
    nombre = os.path.basename(path)
    print(f"Revisando nulos completos en: {nombre}")
    nulos_globales[nombre] = revisar_nulos_completo(path)
nulos_globales["UDP_FLOOD_v2.csv"].head(10)
```

```python
resumen_nulos = []
for nombre, df_nulos in nulos_globales.items():
    total_nulos = df_nulos["nulos"].sum()
    columnas_con_nulos = (df_nulos["nulos"] > 0).sum()
    max_nulos_columna = df_nulos["nulos"].max()
    resumen_nulos.append({
        "dataset": nombre,
        "total_nulos": total_nulos,
        "columnas_con_nulos": columnas_con_nulos,
        "max_nulos_en_una_columna": max_nulos_columna
    })
resumen_nulos_df = pd.DataFrame(resumen_nulos)
display(resumen_nulos_df)
```

Etiquetas unicas

```python
for path in files_paths:
    nombre = os.path.basename(path)
    df = pd.read_csv(
        path,
        header=None,
        names=column_names,
        usecols=["attack_t", "attack_a"],
        nrows=100000
    )
    print("="*60)
    print(nombre)
    print("\nattack_t únicos:")
    print(df["attack_t"].unique())
    print("\nattack_a únicos:")
    print(df["attack_a"].unique())
```

- Todos los subconjuntos contienen 85 atributos.
- No se detectaron valores nulos en los datasets analizados.
- Se evidenció desbalance de clases en distintos niveles según el tipo de ataque.
- Las columnas finales sí corresponden a attack_t y attack_a.
- attack_a es binaria.
- Cada archivo contiene tráfico normal (none) + un tipo específico de ataque.
